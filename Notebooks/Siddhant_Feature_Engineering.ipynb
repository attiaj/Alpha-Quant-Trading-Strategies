```python
import sys
from pathlib import Path
import pandas as pd

ROOT = Path("..").resolve()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from Quantreo.DataPreprocessing import *

data_file = ROOT / "Data" / "FixTimeBars" / "GBPUSD_1H_Demo_READY.csv"

df = pd.read_csv(data_file, index_col="time", parse_dates=True)

print("Rows:", len(df))
print("Columns:", df.columns.tolist())

df.head()
```

```python
df = sma(df, "close", 20)
df = sma(df, "close", 50)
df = rsi(df, "close", 14)
df = bollinger(df, "close", 20)
```

```python
# Simple Moving Averages
df = sma(df, "close", 20)
df = sma(df, "close", 50)

# Relative Strength Index
df = rsi(df, "close", 14)

# Bollinger Bands (20-period, 2 standard deviations)
df = bollinger(df, "close", 20)

print("Updated columns:")
print(df.columns.tolist())

df.tail()
```
